```python
#Layan khalaf 
# Mobile Speed Tests and Hypothesis Testing
# Math 37198, Statistics for Data Science
#1234 

import pandas as pd
import numpy as np
from scipy import stats

# Load the Android data
android_data = pd.read_csv(r'data/android_data_sample.csv')
android_data.head(10)
```

```python
# Question 1 (2 marks): What is the average download speed of android phones in Germany?
# Construct a 95% confidence t-interval for the mean download speed of all android phones in
# Germany.
#filter the data for Germany
android_germany = android_data.loc[ android_data['client_country'] == 'Germany','download_kbps']

# Using the same confidence interval concept as the reference example in week8 Notebook for a population mean
sample_mean = android_germany.mean()
sample_std = android_germany.std()
confidence_level = 0.95
alpha = 1 - confidence_level
n = len(android_germany)
df = n - 1

# Since the population standard deviation is unknown, use the t-distribution
t_critical = stats.t.ppf(1 - alpha / 2, df)
standard_error = sample_std / np.sqrt(n)
margin_of_error = t_critical * standard_error
ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

print(f"Sample size (n): {n}")
print(f"Degrees of freedom (df): {df}")
print(f"Sample mean: {sample_mean:.2f} kbps")
print(f"Sample standard deviation: {sample_std:.2f} kbps")
print(f"Confidence level: {confidence_level * 100:.0f}%")
print(f"Alpha: {alpha:.2f}")
print(f"t-critical value: {t_critical:.4f}")
print(f"Standard error: {standard_error:.2f} kbps")
print(f"Margin of error: {margin_of_error:.2f} kbps")
print(f"95% Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")
print("Interpretation: We are 95% confident that the true mean download speed of all android phones in Germany lies within this interval.")
```

```python
# Question 2 (2 marks): Based on the iPhone data, estimate the population proportion of the data
# that come from the United States and India. Construct a 95% confidence interval for each
# proportion and interpret the results.

# Using the same confidence interval concept as the reference example in week8 Notebook for population proportions
iphone_data = pd.read_csv(r'data/iphone_data_sample.csv')

n = len(iphone_data)
confidence_level = 0.95
alpha = 1 - confidence_level
z_critical = stats.norm.ppf(1 - alpha / 2)

# Sample proportions
p_us = (iphone_data['client_country'] == 'United States').mean()
p_india = (iphone_data['client_country'] == 'India').mean()

# Standard errors for proportions
se_us = np.sqrt((p_us * (1 - p_us)) / n)
se_india = np.sqrt((p_india * (1 - p_india)) / n)

# Margins of error
me_us = z_critical * se_us
me_india = z_critical * se_india

# Confidence intervals
ci_us_lower = p_us - me_us
ci_us_upper = p_us + me_us
ci_india_lower = p_india - me_india
ci_india_upper = p_india + me_india

print(f"Sample size (n): {n}")
print(f"Confidence level: {confidence_level * 100:.0f}%")
print(f"z-critical value: {z_critical:.4f}")

print("\nUnited States")
print(f"Sample proportion: {p_us:.4f}")
print(f"Standard error: {se_us:.4f}")
print(f"Margin of error: {me_us:.4f}")
print(f"95% Confidence Interval: ({ci_us_lower:.4f}, {ci_us_upper:.4f})")
print("Interpretation: We are 95% confident that the true population proportion of iPhone data from the United States lies within this interval.")

print("\nIndia")
print(f"Sample proportion: {p_india:.4f}")
print(f"Standard error: {se_india:.4f}")
print(f"Margin of error: {me_india:.4f}")
print(f"95% Confidence Interval: ({ci_india_lower:.4f}, {ci_india_upper:.4f})")
print("Interpretation: We are 95% confident that the true population proportion of iPhone data from India lies within this interval.")
```

```python
# Question 3 (3 marks): In the Everglades at the Garfield Bight station, the bottom water temperature was
# recorded once a day for 61 days. The average temperature was 30.377 degrees Celsius with a standard
# deviation of 1.7. Use the p-value method to test the claim that the mean temperature is greater than 30
# degrees.

# Using the hypothesis testing method from the reference example in week9 Notebook for a population mean
n = 61
sample_mean = 30.377
sample_std = 1.7
mu_0 = 30
alpha = 0.05
df = n - 1

print(f"H0: mu = {mu_0}")
print(f"Ha: mu > {mu_0}")

# Since sigma is unknown, use a one-sample t-test
standard_error = sample_std / np.sqrt(n)
t_statistic = (sample_mean - mu_0) / standard_error
p_value = 1 - stats.t.cdf(t_statistic, df=df)

print(f"Sample size (n): {n}")
print(f"Sample mean: {sample_mean:.3f}")
print(f"Sample standard deviation: {sample_std:.3f}")
print(f"Standard error: {standard_error:.4f}")
print(f"Test statistic (t): {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Alpha: {alpha}")

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: There is sufficient evidence that the mean bottom water temperature is greater than 30 degrees Celsius.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is not enough evidence that the mean bottom water temperature is greater than 30 degrees Celsius.")
```

```python
# Question 4 (3 marks): The high school athletic director is asked if football players are doing as well
# academically as the other student athletes. We know from a previous study that the average GPA for
# student athletes is 3.10. After an initiative to help improve the GPA of student athletes, the athletic
# director randomly samples 20 football players and finds that the average GPA of the sample is 3.18 with a
# sample standard deviation of 0.54. Use the critical value approach to see if there is a significant
# improvement. Use a 0.05 significance level.

# Using the hypothesis testing method from the reference example in week9 Notebook for a population mean
n = 20
sample_mean = 3.18
sample_std = 0.54
mu_0 = 3.10
alpha = 0.05
df = n - 1

print(f"H0: mu = {mu_0}")
print(f"Ha: mu > {mu_0}")

# Since n < 30 and sigma is unknown, use a one-sample t-test
standard_error = sample_std / np.sqrt(n)
t_statistic = (sample_mean - mu_0) / standard_error
t_critical = stats.t.ppf(1 - alpha, df=df)

print(f"Sample size (n): {n}")
print(f"Sample mean: {sample_mean:.2f}")
print(f"Sample standard deviation: {sample_std:.2f}")
print(f"Standard error: {standard_error:.4f}")
print(f"Test statistic (t): {t_statistic:.4f}")
print(f"Critical value: {t_critical:.4f}")
print(f"Alpha: {alpha}")

if t_statistic > t_critical:
    print("Decision: Reject H0")
    print("Conclusion: There is sufficient evidence of a significant improvement in the GPA of football players.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: There is not enough evidence of a significant improvement in the GPA of football players.")
```
