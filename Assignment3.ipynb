```python
# Assignment 3 
# Layan khalaf 
# 991274007
#Question1: The average time a person spends at a museum is 96 minutes. The    
#standard deviation is 17 minutes. Assume the variable is normally distributed. 
# for Question 1 i used week 5 notes to solve all the problems . 
import scipy.stats
from scipy.stats import norm 
import matplotlib.pyplot as plt
import numpy as np 


```

```python
#a) Generate this data for 500 people and then plot the distribution of including the lines that 
#show:   the mean+/- 1 standard deviation, +/- 2 standard deviations, +/- 3 standard 
#deviations using the given values of µ and 𝜎.  
mean=96
std=17
# genetate data for 500 people 
data = np.random.normal(mean, std, 500)
print(data)

x_axis = np.arange(mean-4*std, mean+4*std, 0.1)
# Mean = 96, SD = 17.
plt.plot(x_axis, norm.pdf(x_axis, mean, std))
plt.axvline(mean)
plt.axvline(mean+std,ls="--",color="red")
plt.axvline(mean-std,ls="--",color="red")
plt.axvline(mean+2*std,ls="--",color="green")
plt.axvline(mean-2*std,ls="--",color="green")
plt.axvline(mean+3*std,ls="--",color="yellow")
plt.axvline(mean-3*std,ls="--",color="yellow")
#68% of the heights are between the red lines, 95% are between the green lines and 99.7% are between the yellow lines
plt.show()
```

```python

#b) If a visitor is selected at random, find the probability that he or she will spend more than 
#120 minutes at the museum.   
b=norm.sf(120,loc=96,scale=17)
print(f"{b:.4f}") 
fig, ax = plt.subplots()
# for distribution curve
x_axis = np.arange(96-4*17, 96+4*17, 0.1)
plt.plot(x_axis, norm.pdf(x_axis, 96, 17))
# for the area under the curve
x_fill = np.arange(120, 96+4*17, 0.1)
plt.fill_between(x_fill, norm.pdf(x_fill, 96, 17), alpha=0.5)
plt.show()


```

```python
#c) If a visitor is selected at random, find the probability that he or she will spend less than 
#80 minutes at the museum. 

c=norm.cdf(80, loc=96, scale=17)
print(f"{c:.4f}")
fig, ax = plt.subplots()
# for distribution curve
x_axis = np.arange(96-4*17, 96+4*17, 0.1)
plt.plot(x_axis, norm.pdf(x_axis, 96, 17))
# for the area under the curve
x_fill = np.arange(96-4*17, 80, 0.1)
plt.fill_between(x_fill, norm.pdf(x_fill, 96, 17), alpha=0.5)
plt.show()
```

```python
#d) If a visitor is selected at random, find the probability that he or she will spend between 90 
#minutes and 110 minutes at the museum. 
##to find if someone is between 90 and 110
below110=norm.cdf(110,loc=96,scale=17)
below90=norm.cdf(90,loc=96,scale=17)
probabilityof90to110=below110-below90
print(f"{probabilityof90to110:.4f}")
#visualize cdf, the shaded area is the probability that you calculted in the previous step
fig, ax = plt.subplots()
 
# for distribution curve
x= np.arange(96-4*17, 96+4*17, 0.001)
ax.plot(x, norm.pdf(x, loc=96, scale=17))
ax.set_title("Normal Dist. with mean=96 minutes, sd=17 minutes")
# for fill_between
px=np.arange(90 ,110, 0.01)
ax.fill_between(px, norm.pdf(px, loc=96, scale=17), 
                alpha=0.5, color='g')
ax.set_ylim(0, 0.045)
plt.show()
```
