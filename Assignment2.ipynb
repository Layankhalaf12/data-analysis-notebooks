```python
# Questio 1B
# Compute z-score of ‘Super Mario 64 DS’ relative to Nintendo games. What can you conclude from this result?
# used the formula we dicussed in class and its mention in Sample Data Frame Basics,Week 1
# Filter Nintendo platforms (e.g., Nintendo )
import pandas as pd
import scipy.stats as st
data = pd.read_csv(r'C:\Users\Layan\OneDrive\Desktop\semster4\MATH37198\Math 37198,Assignment #2\IGN.csv')


nintendo_games = data [data['platform'].str.contains("Nintendo", na=False)]
mario = data[data['title'] == "Super Mario 64 DS"]

# Compute z-score: (value - mean) / std
nintendo_mean = nintendo_games['score'].mean()
nintendo_std = nintendo_games['score'].std()
mario_score = mario['score'].iloc[0]
# used the function directly
mario_zscore = (mario_score - nintendo_mean) / nintendo_std 

print("Conclusion: Super Mario 64 DS  zscore is ", round(mario_zscore, 4), " standard deviation above the mean")
```

```python
# Question 2d
# A PS3 game given that it is Amazing.
# P(PS3 | Amazing) = P(PS3 and Amazing) / P(Amazing)

# Filter rows for PlayStation 3 (platform name in the dataset)
# 
PS3_games = data[data['platform'] == "PlayStation 3"]

# P(PS3 and Amazing)
AmazingAndPS3 = len(PS3_games[PS3_games['score_phrase'] == "Amazing"]) / len(data)

# P(Amazing)
P_Amazing = len(data[data['score_phrase'] == "Amazing"]) / len(data)

# Calculate P(PS3 | Amazing)
Ps3_given_Amazing = round(AmazingAndPS3 / P_Amazing, 4)
Ps3_given_Amazing
```
