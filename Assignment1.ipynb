```python
# Question 1
#In NBA Stats ‘games’ refers to the total number of games played by an individual player or
# team within one or more seasons and is shown as (G).
 
# import library that will be used 
import numpy as np
import pandas as pd
import scipy.stats as st
import matplotlib.pyplot as plt

# read the csv file  NBA stats
df = pd.read_csv(r'C:\Users\Layan\OneDrive\Desktop\semster4\MATH37198\Assignment1\Math 37198,Assignment #1\NBA Stats 2024-25.csv')
df.head()

```

```python
# Question 1A. 
# For the number of games played, find the median and mode.

# assume the number of Game is in column 'G'
games = df['G'] 
# sample median  =  the middle value when the data is sorted 
print(" the  median is ",games.median())
# mode = the value that is most frequent in the data set
print(" the mode is ",games.mode())

```

```python
# Question 1B 
#What is the average number of games played by players under 25 years old? 

# assume the age is in column 'Age'
age = df['Age'] 
# filter the age less than 25 for games played
under25= df.loc[ age < 25 ,'G']
# calculate the average of games played by players under 25
print(np.average(under25))



     
```

```python
# Question 2 
# Free throws are unopposed, one-point shots awarded to players - 
# typically after they are fouled during shooting or when a team exceeds their foul limit. 
# Create a well-labeled histogram of all players’ free throw attempts (FTA). 
# What do you notice, based on your histogram?

# assume the free throw attempts is in column 'FTA'
fta = df['FTA']
# create histogram ,title and labels
plt.hist(fta)
plt.title('Histogram of free throw attempts')
# label x and y axis
plt.xlabel('FTA per game')
plt.ylabel('Number of players')
plt.show()
# the histogram shows that most players have 0-2 free throw attempts per game, while  some players have between 3-5 free throw attempts,
# and very few players have more than 5 free throw attempts.
```

```python
# Question 3b 
# Find the standard deviation of turnovers (TOV) for the Toronto Raptors (TOR).
# Compared to the league-wide standard deviation, 
# are Toronto Raptors players more or less consistent at maintaining possession of the ball?

tov=df['TOV'] 
# filter the Toronto Raptors team
torontoRapters= df[df['Team'] == 'TOR']  
# find the standard deviation
tov_std= tov.std()
tor_std= torontoRapters['TOV'].std()
# print the standard deviation for all players and for the Toronto Raptors
print("  The standard deviation for TOV for all players",tov_std)
print("  The standard deviation for TOV for Toronto Raptors",tor_std) 
# compare the standard deviations
if tor_std < tov_std:
    print("  Toronto Raptors players are more consistent at maintaining possession of the ball")
elif  tor_std > tov_std:
    print("  Toronto Raptors players are less consistent at maintaining possession of the ball")
else:
    print("  Toronto Raptors players are equally consistent at maintaining possession of the ball")
```

```python
# Question 4 
# In basketball, a steal is when a defensive player legally 
# takes the ball away from an opponent. Players who are excellent
# defenders tend to accumulate steals at high rates.
# Find the set of players whose steals per game (STL)
# have a z-score of at least 3.0.
 
# List these players along with their steals per game.
stl = df['STL']
#find z-score of steals per game
z_score = st.zscore(stl)
# list players with at least z score of 3.0 along with their steals per game
players = df.loc[z_score >= 3.0, ['Player', 'STL']]
print(players)


```

```python
# Question 5
#Some teams rely on certain players to take most of their shots. 
# A shot attempt in basketball terms can also be called a field goal attempt. 
# Find the 85th percentile of field goal attempts per game (FGA).
#choose the column FGA
fga= df['FGA'] 
#find the 85th percentile 
P85 = np.quantile(fga, 0.85)
print( "The 85th percentile of field goal attempts per game is", P85)
```
